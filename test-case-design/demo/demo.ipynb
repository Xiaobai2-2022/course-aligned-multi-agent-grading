Using Qwen 3.8 27B and Using `square_root_question.md` to generate a JSON output.


```python
import json

from core.log import log_success, log_fail, log_warning, log_info

from core.io import read_file, write_json

from core.llm import run_qwen3_8
```

```python
prompt = read_file(file_path="../src/test_case_design/prompts/design-test-case.md")

question = read_file(file_path="../src/test_case_design/data/square_root_question.md")

msg = [
    {"role": "system", "content": prompt},
    {"role": "user", "content": question},
]

response = run_qwen3_8(messages=msg, temperature=0.0)

out = json.loads(response) if isinstance(response, str) else response

if not isinstance(out, dict):
    log_fail("The LLM response does not contain a JSON object.")
    raise ValueError("The LLM response must contain a JSON object.")

write_json(data=response, json_path="data/test_cases_out.json")
```
